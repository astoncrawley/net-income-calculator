```python
# Import libraries
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
GROSS_INCOME = 60000
TAX_CODE = '1257L'
```

```python
def personal_allowance(tax_code):
    if tax_code == '1257L':
        return 12570
```

```python
def pay_as_you_earn(gross_income):
    if gross_income <= 12570:
        return 0
    elif 12570 < gross_income <= 50270:
        basic_taxable_income = gross_income - 12570
        return basic_taxable_income * 0.2
    elif 50270 < gross_income <= 125140:
        basic_taxable_income = 50270 - 12570
        higher_taxable_income = gross_income - 50270
        return basic_taxable_income * 0.2 + higher_taxable_income * 0.4
    elif 125140 < gross_income:
        basic_taxable_income = 50270 - 12570
        higher_taxable_income = 125140 - 50270
        additional_taxable_income = gross_income - 125140
        return basic_taxable_income * 0.2 + higher_taxable_income * 0.4 + additional_taxable_income * 0.45
```

```python
def pay_as_you_earn_monthly(monthly_gross_income):
    if monthly_gross_income <= 1048:
        return 0
    elif 1048 < monthly_gross_income <= 4189:
        basic_taxable_income = monthly_gross_income - 1048
        return basic_taxable_income * 0.2
    elif 4189 < monthly_gross_income <= 10428:
        basic_taxable_income = 4189 - 1048
        higher_taxable_income = monthly_gross_income - 4189
        return basic_taxable_income * 0.2 + higher_taxable_income * 0.4
    elif 10428 < monthly_gross_income:
        basic_taxable_income = 4189 - 1048
        higher_taxable_income = 10428 - 4189
        additional_taxable_income = monthly_gross_income - 10428
        return basic_taxable_income * 0.2 + higher_taxable_income * 0.4 + additional_taxable_income * 0.45
```

```python
def national_insurance_contribution(gross_income):
    if gross_income <= 12570:
        return 0
    elif 12570 < gross_income <= 50270:
        basic_taxable_income = gross_income - 12570
        return basic_taxable_income * 0.08
    elif 50270 < gross_income:
        basic_taxable_income = 50270 - 12570
        higher_taxable_income = gross_income - 50270
        return basic_taxable_income * 0.08 + higher_taxable_income * 0.02
```

```python
def national_insurance_contribution_monthly(monthly_gross_income):
    if monthly_gross_income <= 1048:
        return 0
    elif 1048 < monthly_gross_income <= 4189:
        basic_taxable_income = monthly_gross_income - 1048
        return basic_taxable_income * 0.08
    elif 4189 < monthly_gross_income:
        basic_taxable_income = 4189 - 1048
        higher_taxable_income = monthly_gross_income - 4189
        return basic_taxable_income * 0.08 + higher_taxable_income * 0.02
```

```python
def pension_contribution(gross_income):
    if gross_income <= 6240:
        return 0
    elif 6240 < gross_income <= 50270:
        qualifying_earnings = gross_income - 6240
        return qualifying_earnings * 0.05
    elif 50270 < gross_income:
        qualifying_earnings = 50270 - 6240
        return qualifying_earnings * 0.05
```

```python
def pension_contribution_monthly(monthly_gross_income):
    if monthly_gross_income <= 520:
        return 0
    elif 520 < monthly_gross_income <= 4189:
        qualifying_earnings = monthly_gross_income - 520
        return qualifying_earnings * 0.05
    elif 4189 < monthly_gross_income:
        qualifying_earnings = 4189 - 520
        return qualifying_earnings * 0.05
```

```python
def student_loan_repayment(gross_income, loan_plan):
    if loan_plan == 1:
        income_threshold = 26065
        repayment_income = gross_income - income_threshold
        return repayment_income * 0.09
    elif loan_plan == 2:
        income_threshold = 28470
        repayment_income = gross_income - income_threshold
        return repayment_income * 0.09
    elif loan_plan == 4:
        income_threshold = 32745
        repayment_income = gross_income - income_threshold
        return repayment_income * 0.09
    elif loan_plan == 5:
        income_threshold = 25000
        repayment_income = gross_income - income_threshold
        return repayment_income * 0.09
```

```python
def postgraduate_loan_repayment(gross_income):
    income_threshold = 21000
    repayment_income = gross_income - income_threshold
    return repayment_income * 0.06
```

```python
def net_income():
    gross_income = GROSS_INCOME
    tax_code = TAX_CODE
    
    annual_personal_allowance = personal_allowance(tax_code)
    
    annual_pay_as_you_earn = pay_as_you_earn(gross_income)
    monthly_pay_as_you_earn = pay_as_you_earn_monthly(gross_income/12)
    
    annual_national_insurance_contribution = national_insurance_contribution(gross_income)
    monthly_national_insurance_contribution = national_insurance_contribution_monthly(gross_income/12)
    
    annual_pension_contribution = pension_contribution(gross_income)
    monthly_pension_contribution = pension_contribution_monthly(gross_income/12)
    
    annual_student_loan_repayment = student_loan_repayment(gross_income, 2)
    annual_postgraduate_loan_repayment = postgraduate_loan_repayment(gross_income)

    print("ANNUAL OVERVIEW")
    print("Earnings")
    print("Earnings Total: ", gross_income)
    print("")
    print("Deductions")
    print("PAYE: ", annual_pay_as_you_earn)
    print("Employee National Insurance Contribution - A: ", annual_national_insurance_contribution)
    print("NEST (RAS): ", annual_pension_contribution)
    print("Student Loans: ", annual_student_loan_repayment)
    print("Postgraduate Loans: ", annual_postgraduate_loan_repayment)
    print("Deductions Total: ", gross_income)
    print("")
    print("")
    print("MONTHLY OVERVIEW")
    print("Earnings")
    print("Earnings Total: ", gross_income/12)
    print("")
    print("Deductions")
    print("PAYE: ", annual_pay_as_you_earn/12) # Within pennies
    print("PAYE (Monthly): ", monthly_pay_as_you_earn) # Within pennies
    print("Employee National Insurance Contribution - A: ", monthly_national_insurance_contribution) # Accurate
    print("NEST (RAS): ", annual_pension_contribution/12)
    print("NEST (RAS) (Monthly): ", monthly_pension_contribution)
    print("Student Loans: ", math.floor(annual_student_loan_repayment/12)) # £9 underestimate
    print("Postgraduate Loans: ", math.floor(annual_postgraduate_loan_repayment/12)) # Accurate
    print("Deductions Total: ", gross_income)
    
    return 0
```
